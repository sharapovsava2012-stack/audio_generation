```python
from kokoro import KPipeline
from IPython.display import Audio, display
import soundfile as sf
import time

pipeline = KPipeline(lang_code='a')
```

```python
responses = {
    "hello!": "Hello! My name is Alex. Nice to meet you!",
    "hi": "Hi there! I'm Alex. Great to see you!",
    "hey": "Hey! Alex here. How can I help you today?",
    "good morning": "Good morning! This is Alex. Hope you're having a great day!",
    "good afternoon": "Good afternoon! Alex speaking. How's your day going?",
    "good evening": "Good evening! Alex here. Ready to assist you!",
    "what's up": "Not much! Just waiting to chat with you. I'm Alex!",
    "howdy": "Howdy partner! Alex is my name, chatting's my game!",
    "greetings": "Greetings, human! I'm Alex, your friendly AI assistant!",
    "yo": "Yo! Alex in the house! What's cracking?",
    "sup": "Sup! Alex here, ready to rock and roll!",
    "hiya": "Hiya! I'm Alex. So glad you stopped by!",
    "how are you": "I'm doing fantastic, thanks for asking! I'm Alex, by the way!",
    "nice to meet you": "The pleasure is all mine! I'm Alex, your digital companion!"
}

more_responses = {
    "hello there": responses["hello!"],
    "hello": responses["hello!"],
    "heya": responses["hi"],
    "hey there": responses["hey"],
    "good day": responses["good morning"],
    "morning": responses["good morning"],
    "evening": responses["good evening"],
    "wassup": responses["what's up"],
    "whassup": responses["what's up"],
    "howdy there": responses["howdy"],
    "greetings and salutations": responses["greetings"],
    "yo yo": responses["yo"],
    "what's happening": responses["sup"],
    "how's it going": responses["how are you"],
    "how do you do": responses["nice to meet you"],
    "pleased to meet you": responses["nice to meet you"]
}

responses.update(more_responses)
```

```python
def generate_speech(text, voice = 'am_liam', speed =1.0):

  generator = pipeline(text,voice=voice, speed=speed)

  for gs, ps, audio in generator:
    display(Audio(data=audio, rate=24000, autoplay=True))
    time.sleep(0.1)

  return True

def get_response(user_input):
  user_input_lower = user_input.lower().strip()


  if user_input_lower in responses:
    answer = responses[user_input_lower]
    return answer

  else:
    return "Sorry, I dont undeerstand you. Can you please say something from my list? Try saying 'help' to see what I can do!"
```

```python
while True:
    user_input = input("\nВАШ ВОПРОС: ")

    # Проверка на выход
    if user_input.lower() in ['quit', 'exit', 'stop', 'goodbye', 'bye', 'пока']:
        generate_speech("Goodbye! It was nice talking to you!", voice='af_heart')
        break

    # Получаем ответ из словаря
    response_text = get_response(user_input)

    # Генерируем речь
    generate_speech(response_text, voice='am_liam')
```
